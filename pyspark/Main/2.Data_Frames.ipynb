# PySpark DataFrame — Introduction

## 1. What is a DataFrame?

> **A PySpark DataFrame is a distributed collection of data organized into named columns, similar to a table in a relational database or a pandas DataFrame.**

Example:

```text
+---------+---------+-----+
| name    | city    | age |
+---------+---------+-----+
| Karthik | Chennai |  25 |
| Arun    | Mumbai  |  28 |
| Rahul   | Delhi   |  26 |
+---------+---------+-----+
```

A DataFrame has:

```text
Rows
+
Columns
+
Schema
```

---

# 2. RDD vs DataFrame

We already worked with RDDs:

```python
rdd = sc.parallelize([
    ("Karthik", "Chennai", 25),
    ("Arun", "Mumbai", 28)
])
```

With a DataFrame:

```text
name     city       age
Karthik  Chennai    25
Arun     Mumbai     28
```

### Main Difference

**RDD**

```text
Distributed collection of objects
```

**DataFrame**

```text
Distributed tabular data
with named columns and schema
```

For Data Engineering, DataFrames are generally the primary API you'll use.

---

# 3. Creating a DataFrame

The modern entry point is `SparkSession`.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("DataFrameDemo") \
    .getOrCreate()
```

Now create data:

```python
data = [
    ("Karthik", "Chennai", 25),
    ("Arun", "Mumbai", 28),
    ("Rahul", "Delhi", 26)
]
```

Define column names:

```python
columns = ["name", "city", "age"]
```

Create the DataFrame:

```python
df = spark.createDataFrame(data, columns)
```

Display it:

```python
df.show()
```

Output:

```text
+-------+-------+---+
|   name|   city|age|
+-------+-------+---+
|Karthik|Chennai| 25|
|   Arun| Mumbai| 28|
|  Rahul|  Delhi| 26|
+-------+-------+---+
```

---

# 4. `show()`

`show()` displays the DataFrame contents.

```python
df.show()
```

By default, it displays the first 20 rows.

You can specify the number of rows:

```python
df.show(5)
```

You can also display the complete contents of long strings:

```python
df.show(truncate=False)
```

---

# 5. Schema

One of the biggest advantages of DataFrames is that they have a **schema**.

Check the schema:

```python
df.printSchema()
```

Output:

```text
root
 |-- name: string (nullable = true)
 |-- city: string (nullable = true)
 |-- age: long (nullable = true)
```

This tells us:

```text
name → string
city → string
age  → long
```

So DataFrame understands the structure of our data.

---

# 6. `columns`

To get the column names:

```python
df.columns
```

Output:

```python
['name', 'city', 'age']
```

---

# 7. `dtypes`

To get column names along with their data types:

```python
df.dtypes
```

Output:

```python
[
    ('name', 'string'),
    ('city', 'string'),
    ('age', 'bigint')
]
```

---

# 8. `select()`

`select()` is used to choose columns.

```python
df.select("name")
```

Display:

```python
df.select("name").show()
```

Output:

```text
+-------+
|   name|
+-------+
|Karthik|
|   Arun|
|  Rahul|
+-------+
```

Multiple columns:

```python
df.select("name", "city").show()
```

Output:

```text
+-------+-------+
|   name|   city|
+-------+-------+
|Karthik|Chennai|
|   Arun| Mumbai|
|  Rahul|  Delhi|
+-------+-------+
```

---

# 9. `filter()`

Just like RDDs, DataFrames can filter data.

```python
df.filter(df.age > 25).show()
```

Output:

```text
+-----+------+---+
| name|  city|age|
+-----+------+---+
| Arun|Mumbai| 28|
|Rahul| Delhi| 26|
+-----+------+---+
```

Another syntax:

```python
df.filter("age > 25").show()
```

---

# 10. `withColumn()`

`withColumn()` is used to create or modify a column.

For example:

```python
from pyspark.sql.functions import col

df2 = df.withColumn("age_plus_5", col("age") + 5)

df2.show()
```

Output:

```text
+-------+-------+---+----------+
|   name|   city|age|age_plus_5|
+-------+-------+---+----------+
|Karthik|Chennai| 25|        30|
|   Arun| Mumbai| 28|        33|
|  Rahul|  Delhi| 26|        31|
+-------+-------+---+----------+
```

---

DataFrames allow Spark to understand the **structure and meaning of the data**, enabling Spark SQL/Catalyst optimizations.

# Explicit Schema — StructType and StructField

## 1. What is a Schema?

A **schema** defines the structure of a DataFrame.

It tells Spark:

- What columns exist
- What data type each column has
- Whether a column can contain `NULL`

For example:

```text
customers
--------------------------------
name       → String
age        → Integer
salary     → Double
join_date  → Date
```

Instead of asking Spark to infer this structure, we can explicitly tell Spark what it is.

---

# 2. Why Explicit Schema?

Previously, we used:

```python
df = spark.read.csv(
    "customers.csv",
    header=True,
    inferSchema=True
)
```

Here Spark tries to **infer** the data types.

For example:

```text
"25" → Integer
"50000.50" → Double
```

But inference has some disadvantages.

### Problem 1 — Incorrect inference

Suppose a column contains:

```text
00123
00456
00789
```

Spark might interpret it as a number, potentially losing the leading zeros.

But maybe this is actually a:

```text
customer_id
```

that should remain a string.

### Problem 2 — Performance

Schema inference requires Spark to inspect data to determine types.

### Problem 3 — Data quality

Production pipelines usually need a **known contract** for incoming data.

For example:

```text
customer_id → string
age         → integer
salary      → double
```

We don't want the schema unexpectedly changing because today's input data looks different.

---

# 3. `StructType`

`StructType` represents the **complete schema of a DataFrame**.

Think:

```text
StructType
    ↓
Complete DataFrame Schema
```

Import it:

```python
from pyspark.sql.types import StructType
```

Example:

```python
schema = StructType([
    ...
])
```

Inside `StructType`, we define the individual columns using `StructField`.

---

# 4. `StructField`

`StructField` represents **one column definition**.

Think:

```text
StructField
    ↓
One column's definition
```

It contains:

```text
Column Name
+
Data Type
+
Nullable
```

Import:

```python
from pyspark.sql.types import StructField
```

Example:

```python
StructField("name", StringType(), True)
```

This means:

```text
Column name
     ↓
"name"

Data type
     ↓
StringType()

Can contain NULL?
     ↓
True
```

---

# 5. Data Types

PySpark provides many data types.

# 6. Creating an Explicit Schema

Suppose our customer data is:

```python
data = [
    ("C001", "Karthik", 25, 55000.50),
    ("C002", "Arun", 28, 62000.75),
    ("C003", "Rahul", 26, 48000.25)
]
```

We want:

```text
customer_id → string
name        → string
age         → integer
salary      → double
```

Create the schema:

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType,
    DoubleType
)

schema = StructType([
    StructField("customer_id", StringType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("salary", DoubleType(), True)
])
```

Now create the DataFrame:

```python
df = spark.createDataFrame(
    data,
    schema=schema
)
```

Display:

```python
df.show()
```

Output:

```text
+-----------+-------+---+--------+
|customer_id|   name|age|  salary|
+-----------+-------+---+--------+
|       C001|Karthik| 25|55000.50|
|       C002|   Arun| 28|62000.75|
|       C003|  Rahul| 26|48000.25|
+-----------+-------+---+--------+
```

Check:

```python
df.printSchema()
```

Output:

```text
root
 |-- customer_id: string (nullable = true)
 |-- name: string (nullable = true)
 |-- age: integer (nullable = true)
 |-- salary: double (nullable = true)
```

---

# 7. Understanding `nullable`

Look at:

```python
StructField(
    "age",
    IntegerType(),
    True
)
```

The third argument is:

```text
True
```

This means the column is allowed to contain `NULL`.

For example:

```python
StructField("age", IntegerType(), True)
```

means:

```text
age
 ↓
Integer
 ↓
NULL allowed
```

If we use:

```python
StructField("age", IntegerType(), False)
```

it means:

```text
age
 ↓
Integer
 ↓
NULL not allowed
```

So:

```python
StructField("column_name", DataType(), nullable)
```

---

# 8. Complete Example

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType,
    DoubleType
)

spark = SparkSession.builder \
    .appName("ExplicitSchema") \
    .getOrCreate()

data = [
    ("C001", "Karthik", 25, 55000.50),
    ("C002", "Arun", 28, 62000.75),
    ("C003", "Rahul", 26, 48000.25)
]

schema = StructType([
    StructField("customer_id", StringType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("salary", DoubleType(), True)
])

df = spark.createDataFrame(
    data,
    schema=schema
)

df.show()

df.printSchema()
```

# 9. Explicit Schema with CSV

This is where explicit schemas become particularly useful in Data Engineering.

Instead of:

```python
df = spark.read.csv(
    "customers.csv",
    header=True,
    inferSchema=True
)
```

we can define:

```python
schema = StructType([
    StructField("customer_id", StringType(), False),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("salary", DoubleType(), True)
])
```

Then:

```python
df = spark.read.csv(
    "customers.csv",
    header=True,
    schema=schema
)
```

Now Spark doesn't need to guess the types.

The pipeline becomes:

```text
CSV
 ↓
Explicit Schema
 ↓
DataFrame
```

# 10. Why This Matters in Real Data Engineering

Imagine a production CSV arrives every day:

```text
customers_2026_10_01.csv
customers_2026_10_02.csv
customers_2026_10_03.csv
```

You expect:

```text
customer_id → string
name        → string
age         → integer
salary      → double
```

If you use schema inference, Spark determines the types from the incoming data.

If you explicitly define the schema:

```text
                Expected Schema
                       ↓
CSV Data → Schema Validation → DataFrame
```

This gives your pipeline a predictable structure.

This is especially important when working with:

- ETL pipelines
- Data Lakes
- Databricks
- Delta Lake
- Production batch jobs
- Streaming pipelines
